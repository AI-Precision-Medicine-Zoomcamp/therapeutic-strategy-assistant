# 08 - DrugBank EGFR Enrichment

Goal: check whether DrugBank can be used for EGFR enrichment in this project.

DrugBank is optional for the MVP. If academic downloads are unavailable or licensing is unclear, the project should continue with ChEMBL, DGIdb, openFDA, Open Targets, PubMed, ClinicalTrials.gov, UniProt, and DrugCentral.

Output: `egfr_drugbank_access_status.csv`

### 1. Test notebook environment

```python
import sys
from pathlib import Path

import pandas as pd
import requests

print("Notebook is working")
print("Python executable:", sys.executable)
```

### 2. Set project folders

```python
PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()

DATA_DIR = PROJECT_ROOT / "data"
RAW_DIR = DATA_DIR / "raw" / "drugbank"
PROCESSED_DIR = DATA_DIR / "processed"

RAW_DIR.mkdir(parents=True, exist_ok=True)
PROCESSED_DIR.mkdir(parents=True, exist_ok=True)

print("Project root:", PROJECT_ROOT)
print("Raw data folder:", RAW_DIR)
print("Processed data folder:", PROCESSED_DIR)
```

### 3. Check DrugBank release/download page

```python
target_name = "EGFR"
drugbank_release_url = "https://go.drugbank.com/releases/latest"

response = requests.get(drugbank_release_url, timeout=(10, 60))
print("HTTP status:", response.status_code)

raw_file = RAW_DIR / "drugbank_latest_release_page.html"
raw_file.write_text(response.text)

print("Saved raw release page:", raw_file)
```

### 4. Build DrugBank access status record

```python
page_text = response.text.lower()

academic_downloads_paused = "academic downloads are temporarily paused" in page_text or "temporarily unavailable" in page_text
access_status = "temporarily_unavailable" if academic_downloads_paused else "check_required"

status_row = {
    "target_name": target_name,
    "source": "DrugBank",
    "source_url": drugbank_release_url,
    "access_status": access_status,
    "required_for_mvp": False,
    "planned_processed_outputs": "egfr_drugbank_targets.csv | egfr_drugbank_summary.csv",
    "reason_not_required": "Existing public/API-backed sources are sufficient for the EGFR MVP.",
    "recommended_action": "Do not block MVP on DrugBank. Add later only if access and licensing are confirmed.",
}

drugbank_status_df = pd.DataFrame([status_row])
drugbank_status_df
```

### 5. Save processed DrugBank status

```python
status_file = PROCESSED_DIR / "egfr_drugbank_access_status.csv"
drugbank_status_df.to_csv(status_file, index=False)

print("Saved:", status_file)
print("Rows:", len(drugbank_status_df))
```

### 6. Final result

```python
print("DrugBank Access Check Complete")
print("=" * 70)
print("Target:", target_name)
print("Access status:", access_status)
print("Required for MVP:", False)
print("Processed file:", status_file)
display(drugbank_status_df)
```
